# Conversion XLSX → CSV avec UTF-8 et conservation de formats

## Méthode 1 : Conversion simple avec Pandas (UTF-8 + accents)
La façon la plus simple - les accents sont préservés automatiquement

```python
# Import des packages
import pandas as pd
```

```python
# Lecture du XLSX
df = pd.read_excel("neologie - description physique, forme, couleurs - microlearning.xlsx", sheet_name=0)

# Conversion en CSV avec UTF-8
df.to_csv("neologie-description-physique-forme-couleurs-microlearning.csv", index=False, encoding='utf-8', sep=';')

print("✓ Fichier converti avec succès en UTF-8")
print(df)
```

## Méthode 2 : Conversion multi-feuilles (si plusieurs sheets)

```python
# Lire toutes les feuilles
excel_file = pd.ExcelFile("votre_fichier.xlsx")

# Convertir chaque feuille en CSV
for sheet_name in excel_file.sheet_names:
    df = pd.read_excel("votre_fichier.xlsx", sheet_name=sheet_name)
    output_name = f"{sheet_name}.csv"
    df.to_csv(output_name, index=False, encoding='utf-8')
    print(f"✓ {sheet_name} → {output_name}")
    
print(f"\nFeuilles trouvées: {excel_file.sheet_names}")
```

## Méthode 3 : Conversion avec préservation de formats (couleurs, bold, italic)
**Note** : Les formats ne peuvent pas être sauvegardés en CSV pur. Cette méthode crée un CSV enrichi avec métadonnées HTML optionnelles.

```python
from openpyxl import load_workbook
import csv

def convert_xlsx_with_formats(xlsx_path, csv_path):
    """Convertit XLSX en CSV en préservant les types de données et formats basiques"""
    
    wb = load_workbook(xlsx_path)
    ws = wb.active
    
    with open(csv_path, 'w', newline='', encoding='utf-8') as csvfile:
        writer = csv.writer(csvfile)
        
        for row in ws.iter_rows(values_only=False):
            csv_row = []
            for cell in row:
                value = cell.value
                
                # Appliquer la mise en forme si présente
                if value is not None:
                    # Préserver les nombres, dates, etc.
                    if cell.data_type == 'n':  # Nombre
                        csv_row.append(str(value))
                    elif cell.data_type == 'd':  # Date
                        csv_row.append(str(value))
                    else:
                        csv_row.append(str(value))
                else:
                    csv_row.append('')
            
            writer.writerow(csv_row)

# Utilisation
convert_xlsx_with_formats("votre_fichier.xlsx", "votre_fichier.csv")
print("✓ Conversion avec formats complétée")
```

## Résumé des options

| Option | Avantages | Inconvénients |
|--------|-----------|---------------|
| **Méthode 1** | Simple, rapide, UTF-8 natif, accents préservés | Formats visuels perdus (couleurs, fonts) |
| **Méthode 2** | Gérer plusieurs feuilles | Formats visuels perdus |
| **Méthode 3** | Contrôle fin des types de données | Format CSV limité de nature |

**Important** : Le format CSV est texte pur. Les couleurs, polices et mise en forme ne peuvent pas être conservées. Seul le contenu (texte, nombres, dates) est préservé.
